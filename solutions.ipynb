```python
import pandas as pd

df = pd.read_csv('Robintrack_subsample.csv')

numeric_cols = ['users_lag', 'Users', 'minutes', 'delay', 'random']

stats = pd.DataFrame({
    'Variable': numeric_cols,
    'N':    [df[c].count()         for c in numeric_cols],
    'Mean': [df[c].mean()          for c in numeric_cols],
    'SD':   [df[c].std()           for c in numeric_cols],
    'Min':  [df[c].min()           for c in numeric_cols],
    'p25':  [df[c].quantile(0.25)  for c in numeric_cols],
    'p50':  [df[c].quantile(0.50)  for c in numeric_cols],
    'p75':  [df[c].quantile(0.75)  for c in numeric_cols],
    'Max':  [df[c].max()           for c in numeric_cols],
}).set_index('Variable')

stats
```
